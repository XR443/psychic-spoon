1. Приведите примеры кода, как можно создать, обновить, удалить или провести какое-нибудь действие (конвертация, извлечение фич) над файлом (можно без реализации деталей).


````python
from media import Photo, Storage, Media


class MyStorage(Storage):

    def __init__(self):
        self.storage = dict()
        self.__sequence = 0

    def save(self, media: Media):
        sequence = self.__sequence
        self.storage[sequence] = media
        self.__sequence += 1
        return sequence

    def load(self, id) -> Media:
        return self.storage[id]


media = Photo("file_name", "me")
storage = MyStorage()

id_of_saved_media = storage.save(media)

loaded_media_by_id = storage.load(id_of_saved_media)

assert media == loaded_media_by_id
````
