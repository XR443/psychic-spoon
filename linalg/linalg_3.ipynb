#### Дано:
$ A = \begin{pmatrix} 1 & 2 & 3 \\ 3 & 6 & p \\ 5 & 10 & q \end{pmatrix}$

#### Найти:
Такие $p$ и $q$ при которых $\mathrm{rang}(A) = 1$ 

#### Решение

Ранг матрицы равен рангу наибольшего из его миноров.

Для начала необходимо убедиться что определители миноров второго порядка, в которых нет неизвестных равны 0.

$\begin{vmatrix} 1 & 2 \\ 3 & 6 \end{vmatrix} = 1 * 6 - 2 * 3 = 6 - 6 = 0$

$\begin{vmatrix} 3 & 6 \\ 5 & 10 \end{vmatrix} = 3 * 10 - 6 * 5 = 30 - 30 = 0$

Определители миноров второго порядка, в которых нет неизвестных, равны нулю.

Чтобы у матрицы был ранг равен 1, необходимо чтобы определители двух последних миноров второго порядка равнялись нулю.

$\begin{vmatrix} 2 & 3 \\ 6 & p \end{vmatrix} = 2 * p - 3 * 6 = 2p - 18 = 0$

$\begin{vmatrix} 6 & p \\ 10 & q \end{vmatrix} = 6 * q - p * 10 = 6q - 10p = 0$

$
\begin{cases}
2p - 18 = 0 \\
6q - 10p = 0 \\
\end{cases} 
\implies
\begin{cases}
2p = 18 \\
6q - 10p = 0 \\
\end{cases} 
\implies
\begin{cases}
p = 9 \\
6q - 10p = 0 \\
\end{cases} 
$

$6q - 10p = 0$

$6q - 10 * 9 = 0$

$6q = 90$ 

$q = \dfrac{90}{6} = \dfrac{30}{2} = 15$ 

Подставим полученные значения в:

$\begin{vmatrix} 2 & 3 \\ 6 & p \end{vmatrix} = 2 * p - 3 * 6 = 2p - 18 = 2 * 9 - 18 = 18 - 18 = 0$

$\begin{vmatrix} 6 & p \\ 10 & q \end{vmatrix} = 6 * q - p * 10 = 6q - 10p = 6 * 15 - 10 * 9 = 90 - 90 = 0$

Определители равны 0

#### Ответ:
$
\begin{cases}
q = 9 \\
p = 15 \\
\end{cases} 
$

#### Альтернативное решение:

При рассмотрении матрицы можно заметить, что столбец 2 является линейной комбинацией столбца 1 умноженного на 2.

Значение первой строки третьего столбца является значением первой строки первого столбца умноженного на 3.

Из этого можно сделать вывод, что что значения $p$ и $q$ можно рассчитать умножив первый столбец на 3.

$\begin{pmatrix} 1 \\ 3 \\ 5 \end{pmatrix} * 3 = \begin{pmatrix} 3 \\ 9 \\ 15 \end{pmatrix}$

Мы получили такие же значения $p$ и $q$ как и в при решении через определители.

Так же мы доказали, что столбцы 2 и 3 являются линейной комбинацией стобца 1 $\implies$ ранг матрицы не может быть равен 3, т.к. по свойсву матриц её определитель равен нулю.
