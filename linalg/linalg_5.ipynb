#### Дано:

$A \in \mathbb{R}^{n \times n}, n \geq 100$

- Выполнить SVD разложение
- Построить график зависимости ошибки аппроксимации матрицы от её ранга

#### NumPy:

```python
import numpy as np
import matplotlib.pyplot as plt

n = 100
np.random.seed(42)
A = np.random.rand(n, n)

U, S, Vh = np.linalg.svd(A)

k = 1

rang_2_norm = dict()

for k in range (1,101):
    U_k = U[:, :k]
    S_k = S[:k]
    Vh_k = Vh[:k, :]

    A_k = U_k @ np.diag(S_k) @ Vh_k
    fro_norm = np.linalg.norm(A - A_k, 'fro')
    
    rang_2_norm[k] = fro_norm
    
x = rang_2_norm.keys()
y = rang_2_norm.values()

fig, ax = plt.subplots()

ax.plot(x, y)

plt.title("Зависимость ошибки аппроксимации от ранга аппроксимированной матрицы")
plt.xlabel("Ранг аппроксимированной матрицы")
plt.ylabel("Ошибка аппроксимации")

plt.show()
```
