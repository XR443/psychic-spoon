#### Дано:
$S = \begin{Bmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} & \begin{pmatrix} 0 \\ 1 \end{pmatrix} \end{Bmatrix}$

$a_1 = \begin{pmatrix} 2 & -5 \end{pmatrix}^\top$

$a_2 = \begin{pmatrix} -1 & 3 \end{pmatrix}^\top$

$x = \begin{pmatrix} 1 & -4 \end{pmatrix}^\top$

$\begin{bmatrix} y \end{bmatrix}_\mathbb{B}  = \begin{pmatrix} 1 & 1 \end{pmatrix}^\top$

##### Доп. инфо:
Разница линейной комбинации линейно-независимых векторов от одного базиса дает нулевой вектор только в случае если их скаляры представить равными нулю. 

#### Доказать линейную независимость векторов $a_1$ и $a_2$:

#### Найти:

- Координаты вектора $x$ в базисе $\mathbb{B}$

- Координаты вектора $y$ в стандартном базисе $S$

#### Решение

- ##### Доказательство независимости векторов $a_1$ и $a_2$ :

Проверим являются ли вектора $a_1$ и $a_2$ линейно-независимыми.

$\mathbb{0} = \gamma_1 a_1 + \gamma_2 a_2$

$\mathbb{0} = \gamma_1 \begin{pmatrix} 2 & -5 \end{pmatrix} + \gamma_2 \begin{pmatrix} -1 & 3 \end{pmatrix}$

$\mathbb{0} = \begin{pmatrix} \gamma_1 2 - \gamma_2 & \gamma_1 (-5) + \gamma_2 3 \end{pmatrix}$

$
\begin{cases}
\gamma_1 2 - \gamma_2 = 0 \\
\gamma_1 (-5) + \gamma_2 3 = 0 \\
\end{cases} 
$

$\begin{pmatrix} 2 & -1 & | & 0 \\ -5 & 3 & | & 0 \end{pmatrix}$

Узнаем совместна ли система линейных уравнений. Для этого необходимо найти их ранг.

Дополненный столбец равен нулю, по свойству матриц определитель минора с нулевым столбцом равен нулю, поэтому рассматривать будем следующий минор:
$\begin{pmatrix} 2 & -1 \\ -5 & 3 \end{pmatrix}$

$\begin{vmatrix} 2 & -1 \\ -5 & 3 \end{vmatrix} = 2 * 3 - (-1) * (-5) = 6 - 5 = 1$

Определитель минора не равен нулю $\implies$ Ранг основной матрицы равен рангу расширенной матрицы $\implies$ Система совместна и определенна т.к. ранг матрицы равен числу переменных.

Найдем методом Крамера переменные:

$\gamma_1 = \dfrac{\begin{vmatrix} 0 & -1 \\ 0 & 3 \end{vmatrix}}{\begin{vmatrix} 2 & -1 \\ -5 & 3 \end{vmatrix}} = \dfrac{0}{1} = 0$

$\gamma_2 = \dfrac{\begin{vmatrix} 2 & 0 \\ -5 & 0 \end{vmatrix}}{\begin{vmatrix} 2 & -1 \\ -5 & 3 \end{vmatrix}} = \dfrac{0}{1} = 0$

$\gamma_1 = 0$ и $\gamma_2 = 0$ $\implies$ вектора $a_1$ и $a_2$ линейно независимы   

- ##### Нахождение координат вектора $x$ в базисе $\mathbb{B}$

$\begin{bmatrix} x \end{bmatrix}_S = \begin{pmatrix} 1 \\ -4 \end{pmatrix}$

$
\begin{bmatrix} x \end{bmatrix}_S = B * \begin{bmatrix} x \end{bmatrix}_\mathbb{B}
\implies
B^{-1} * \begin{bmatrix} x \end{bmatrix}_S = B^{-1} * B * \begin{bmatrix} x \end{bmatrix}_\mathbb{B}
\implies
B^{-1} * \begin{bmatrix} x \end{bmatrix}_S = E * \begin{bmatrix} x \end{bmatrix}_\mathbb{B}
\implies
B^{-1} * \begin{bmatrix} x \end{bmatrix}_S = \begin{bmatrix} x \end{bmatrix}_\mathbb{B}
$

$\begin{pmatrix} 3 & 1 \\ 5 & 2 \end{pmatrix} *  \begin{pmatrix} 1 \\ -4 \end{pmatrix} = \begin{pmatrix} -1 \\ -3 \end{pmatrix}$

Проверка:

$B * \begin{bmatrix} x \end{bmatrix}_\mathbb{B} = \begin{bmatrix} x \end{bmatrix}_S$

$\begin{pmatrix} 2 & -1 \\ -5 & 3 \end{pmatrix} *  \begin{pmatrix} -1 \\ -3 \end{pmatrix} = \begin{pmatrix} 1 \\ -4 \end{pmatrix} $

- ##### Нахождение координат вектора $y$ в базисе $S$

$\begin{bmatrix} y \end{bmatrix}_\mathbb{B} = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$

$\begin{bmatrix} y \end{bmatrix}_S = B * \begin{bmatrix} y \end{bmatrix}_\mathbb{B}$

$\begin{pmatrix} 2 & -1 \\ -5 & 3 \end{pmatrix} *  \begin{pmatrix} 1 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ -2 \end{pmatrix} $

Проверка:

$B^{-1} * \begin{bmatrix} y \end{bmatrix}_S = \begin{bmatrix} x \end{bmatrix}_\mathbb{B}$

$\begin{pmatrix} 3 & 1 \\ 5 & 2 \end{pmatrix} *  \begin{pmatrix} 1 \\ -2 \end{pmatrix} = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$

#### Ответ:

Вектора $a_1$ и $a_2$ линейно независимы.

$\begin{bmatrix} x \end{bmatrix}_\mathbb{B} = \begin{pmatrix} -1 \\ -3 \end{pmatrix}$

$\begin{bmatrix} y \end{bmatrix}_s = \begin{pmatrix} 1 \\ -2 \end{pmatrix}$

#### NumPy проверка:

```python
import numpy as np

a_1 = np.matrix([[2], [-5]])
a_2 = np.matrix([[-1], [3]])

B = np.concatenate((a_1.transpose(), a_2.transpose())).transpose()
xS = np.matrix([[1], [-4]])
```

```python
expected_xB = np.matrix([[-1.], [-3.]])
actual_xB = np.linalg.inv(B) * xS
assert np.isclose(expected_xB, actual_xB).all()
```

```python
expected_xS = np.matrix([[1.], [-4.]])
actual_xS = B * (np.linalg.inv(B) * xS)
assert np.isclose(expected_xS, actual_xS).all()
```

```python
expected_yS = np.matrix([[1.], [-2.]])
actual_yS = B * np.matrix([[1], [1]])
assert np.isclose(expected_yS, actual_yS).all()
```
