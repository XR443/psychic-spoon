#### Дано:
$A = \begin{pmatrix} 2 & -4 \\ 3 & 5 \\ -1 & 0 \end{pmatrix}$, 
$B = \begin{pmatrix} 1 & 2 & 7 \\ -3 & -4 & 0 \\ 5 & 2 & 1 \end{pmatrix}$, 
$C = \begin{pmatrix} 6 & -3 & 9 \\ 4 & -5 & 2 \\ 8 & 1 & 5 \end{pmatrix}$.

#### Найти:
$D = A^\top C - 2A^\top B^\top$

#### Решение

$A^\top = \begin{pmatrix} 2 & 3 & -1 \\ -4 & 5 & 0 \end{pmatrix}$, 
$B^\top = \begin{pmatrix} 1 & -3 & 5 \\ 2 & -4 & 2 \\ 7 & 0 & 1 \end{pmatrix}$ 

$X = A^\top C = \begin{pmatrix} 
(2 * 6 + 3 * 4 + (-1) * 8) & (2 * (-3) + 3 * (-5) + (-1) * 1) & (2 * 9 + 3 * 2 + (-1) * 5 ) 
\\ ((-4) * 6 + 5 * 4) & ((-4) * (-3) + 5 * (-5)) & ((-4) * 9 + 5 * 2)
\end{pmatrix}
= \begin{pmatrix} 
(12 + 12 - 8) & (-6 - 15 - 1) & (18 + 6 - 5) 
\\ (-24 + 20) & (12 - 25) & (-36 + 10)
\end{pmatrix}
= \begin{pmatrix} 
16 & -22 & 19
\\ -4 & -13 & -26
\end{pmatrix}$

$Y = 2A^\top B^\top = 2\begin{pmatrix} 
(2 * 1 + 3 * 2 + (-1) * 7) & (2 * (-3) + 3 * (-4)) & (2 * 5 + 3 * 2 + (-1) * 1) 
\\ ((-4) * 1 + 5 * 2) & ((-4) * (-3) + 5 * (-4)) & ((-4) * 5 + 5 * 2)
\end{pmatrix}
= \begin{pmatrix} 
(2 + 6 - 7) & (-6 - 12) & (10 + 6 - 1) 
\\ (-4 + 10) & (12 - 20) & (-20 + 10)
\end{pmatrix}
= 2\begin{pmatrix} 
1 & -18 & 15
\\ 6 & -8 & -10
\end{pmatrix}
= \begin{pmatrix} 
2 & -36 & 30
\\ 12 & -16 & -20
\end{pmatrix}$

$D = \begin{pmatrix} 
16 & -22 & 19
\\ -4 & -13 & -26
\end{pmatrix} - \begin{pmatrix} 
2 & -36 & 30
\\ 12 & -16 & -20
\end{pmatrix}
= \begin{pmatrix} 
16 - 2 & -22 - (-36) & 19 - 30
\\ -4 - 12 & -13 - (-16) & -26 - (-20)
\end{pmatrix}
= \begin{pmatrix} 
14 & 14 & -11
\\ -16 & 3 & -6
\end{pmatrix}$

#### Ответ:
$D = \begin{pmatrix} 14 & 14 & -11 \\ -16 & 3 & -6 \end{pmatrix}$

#### NumPy проверка:

```python
import numpy as np

A = np.matrix([[2, -4], [3, 5], [-1, 0]])
B = np.matrix([[1,  2, 7], [-3, -4, 0], [5, 2, 1]])
C = np.matrix([[6, -3, 9], [4, -5, 2], [8, 1, 5]])

X = A.transpose() * C
Y = 2 * A.transpose() * B.transpose()

R_expected = np.matrix([[14, 14, -11], [-16, 3, -6]])
R_actual = X - Y

assert (R_expected & R_actual).all()
```
