#### Дано:

$f(x) = \frac{1}{3}\begin{Vmatrix} x \end{Vmatrix}^3_2$, $x \in \mathbb{R}^n \backslash \{0\}$

#### Найти:

$df(x)$

$\nabla f(x)$

#### Решение

$f(x) = \frac{1}{3}\begin{Vmatrix} x \end{Vmatrix}^3_2$

$\begin{Vmatrix} x \end{Vmatrix}^3_2 = \sqrt{\sum_i |x|^2}^3 = \sqrt{<x, x>}^3$

$\frac{1}{3} \sqrt{<x, x>}^3 \rightarrow min$

$
dF(x) = d(\frac{1}{3} \sqrt{<x, x>}^3) 
= d(\frac{1}{3} <x, x>^\frac{3}{2}) \\
= \frac{1}{3} d(<x, x>^\frac{3}{2}) \\
= \frac{1}{3} d(<x, x>^\frac{3}{2}) \cdot d(<x, x>) \\
= \frac{1}{3} \frac{3}{2} <x, x>^\frac{1}{2} \cdot d(<x, x>) \\
= \frac{1}{3} \frac{3}{2} <x, x>^\frac{1}{2} \cdot (<dx, x> + <x, dx>) \\
= \frac{1}{3} \frac{3}{2} <x, x>^\frac{1}{2} \cdot 2<x, dx> \\
= <x, x>^\frac{1}{2} \cdot <x, dx> \\
= \sqrt{<x, x>} \cdot <x, dx> \\
= \begin{Vmatrix} x \end{Vmatrix}_2 \cdot <x, dx> \\
= \begin{Vmatrix} x \end{Vmatrix}_2 \cdot x
$

#### Ответ:

$\begin{Vmatrix} x \end{Vmatrix}_2 \cdot x$
