#### Дано:

$f(x) = -e^{-x^\top x}$, $x \in \mathbb{R}^n$

#### Найти:

$df(x)$

$\nabla f(x)$

#### Решение

$f(x) = -e^{-x^\top x} = -e^{-<x,x>}$

$-e^{-<x,x>} \rightarrow min$

$
dF(x) = d(-<x,x>) \cdot d(-e^{-<x,x>}) = -(<dx,x>+<x,dx>) \cdot -e^{-<x,x>}
= -2(<x, dx>) \cdot -e^{-<x,x>}
= -2x \cdot -e^{-<x,x>}
= 2x \cdot e^{-<x,x>}
= \frac{2x}{e^{<x,x>}}
$

#### Ответ:

$\frac{2x}{e^{<x,x>}}$
