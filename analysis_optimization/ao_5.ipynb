#### Дано:

$f(x) = \begin{Vmatrix} x \end{Vmatrix}_2$, $x \in \mathbb{R}^n \backslash \{0\}$

#### Найти:

$df(x)$

$\nabla f(x)$

#### Решение

$f(x) = \begin{Vmatrix} x \end{Vmatrix}_2$

$\begin{Vmatrix} x \end{Vmatrix}_2 = \sqrt{\sum_i |x|^2} = \sqrt{<x, x>}$

$\sqrt{<x, x>} \rightarrow min$

$
dF(x) = \sqrt{<x, x>} 
= d(<x, x>^\frac{1}{2}) \\
= \frac{1}{2} \frac{1}{<x, x>^\frac{1}{2}} d(<x,x>) \\
= \frac{1}{2} \frac{1}{\sqrt{<x, x>}} d(<x,x>) \\
= \frac{1}{2} \frac{1}{\sqrt{<x, x>}} \cdot (<dx,x> + <x,dx>) \\
= \frac{1}{2} \frac{1}{\sqrt{<x, x>}} \cdot 2<x, dx> \\
= \frac{1}{\sqrt{<x, x>}} \cdot <x, dx> \\
= \frac{<x, dx>}{\sqrt{<x, x>}} \\
= \frac{x}{\sqrt{<x, x>}} \\
= \frac{x}{{||x||}_2} \\
$

#### Ответ:

$\frac{x}{{||x||}_2}$
