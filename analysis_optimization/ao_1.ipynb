#### Дано:
$f(x) = x^3_1 - 2x_1x_2 + x^2_2 - 3x_1 - 2x_2, x \in \mathbb{R^2}$

#### Найти:

$x_c$ при которых $\nabla f(x_c) = 0$

#### Решение

Найдем частную производную по $x_1$

$\frac{\partial F_{x_1}'((x_1, x_2))}{\partial x_1} = x^3_1 - 2x_1x_2 + x_2^2 - 3x_1 - 2x_2 = 3x_1^2 - 2x_2 - 3$


Найдем частную производную по $x_2$

$\frac{\partial F_{x_2}'((x_1, x_2))}{\partial x_2} = x^3_1 - 2x_1x_2 + x^2_2 - 3x_1 - 2x_2 = -2x_1 + 2x_2 - 2$

Тогда:

$\nabla f(x) = \begin{Bmatrix} \frac{\partial F}{\partial x_1} & \frac{\partial F}{\partial x_2} \end{Bmatrix} 
= \begin{Bmatrix} (3x_1^2 - 2x_2 - 3) & (-2x_1 + 2x_2 - 2) \end{Bmatrix}$

$
\begin{cases}
3x_1^2 - 2x_2 - 3 = 0 \\
-2x_1 + 2x_2 - 2 = 0 \\
\end{cases}
$

Разделим второе уравнение на 2:

$
\begin{cases}
3x_1^2 - 2x_2 - 3 = 0 \\
-x_1 + x_2 - 1 = 0 \\
\end{cases}
\implies
\begin{cases}
3x_1^2 - 2x_2 - 3 = 0\\
x_2 = x_1 + 1 \\
\end{cases}
\implies
3x_1^2 - 2(x_1 + 1) - 3 = 0
$


Решим квадратное уравнение: $3x_1^2 - 2(x_1 + 1) - 3 = 0$

$
3y^2 - 2(y + 1) - 3 = 0 \\
3y^2 - 2y - 2 - 3 = 0 \\
3y^2 - 2y - 5 = 0 \\
$

$D = (-2)^2 - 4 \cdot 3 \cdot (-5) = 4 + 60 = 64$

$y_1 = \frac{2 - \sqrt{64}}{2 \cdot 3} = \frac{2 - 8}{6} = \frac{-6}{6} = -1$

$y_2 = \frac{2 - \sqrt{64}}{2 \cdot 3} = \frac{2 + 8}{6} = \frac{10}{6} = \frac{5}{3}$

$x_1 = \begin{cases} -1 \\ \frac{5}{3} \end{cases}$

$x_2 = x_1 + 1 = \begin{cases} x_2 = -1 + 1 \\ x_2 = \frac{5}{3} + 1 \end{cases} = \begin{cases} x_2 = 0 \\ x_2 = \frac{8}{3} \end{cases}$

#### Ответ:

$x_c = \begin{pmatrix} -1 \\ 0 \end{pmatrix}, \begin{pmatrix} \frac{5}{3} \\ \frac{8}{3} \end{pmatrix}$
