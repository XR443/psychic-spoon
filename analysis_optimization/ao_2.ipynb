#### Дано:

$f(x) = \ln(\sqrt{x_1} + \sqrt{x_2})$

#### Найти:

$x_1 \frac{\partial f}{\partial x_1} +  x_2 \frac{\partial f}{\partial x_2} = \frac{1}{2}$

#### Решение

$f(x) = \ln(\sqrt{x_1} + \sqrt{x_2})$

$g(b) = \sqrt{b_1} + \sqrt{b_2}$

$h(a) = \ln(g(a))$

$f(x) = h(x)$

Найдем частную производную по $x_1$:

$g'(b) = \sqrt{b_1}' + \sqrt{b_2}' = b_1^{\frac{1}{2}} = \frac{1}{2} \cdot b_1^{-\frac{1}{2}} = \frac{1}{2 \sqrt{b_1}}$

$h'(a) = \ln(g(a))' = \ln(\sqrt{a_1} + \sqrt{a_2})' = \frac{1}{\sqrt{a_1} + \sqrt{a_2}}$

$
f'(x) = h'(x) = \ln(g(x))' \cdot g'(x) 
= \frac{1}{\sqrt{x_1} + \sqrt{x_2}} \cdot \frac{1}{2 \sqrt{x_1}} 
= \frac{1}{2 \sqrt{x_1}(\sqrt{x_1} + \sqrt{x_2})}
= \frac{1}{ 2x_1 + 2\sqrt{x_1 x_2}}
$


Найдем частную производную по $x_2$:

$g'(b) = \sqrt{b_1}' + \sqrt{b_2}' = b_2^{\frac{1}{2}} = \frac{1}{2} \cdot b_2^{-\frac{1}{2}} = \frac{1}{2 \sqrt{b_2}}$

$h'(a) = \ln(g(a))' = \ln(\sqrt{a_1} + \sqrt{a_2})' = \frac{1}{\sqrt{a_1} + \sqrt{a_2}}$

$
f'(x) = h'(x) = \ln(g(x))' \cdot g'(x) 
= \frac{1}{\sqrt{x_1} + \sqrt{x_2}} \cdot \frac{1}{2 \sqrt{x_2}} 
= \frac{1}{2 \sqrt{x_2}(\sqrt{x_1} + \sqrt{x_2})}
= \frac{1}{ 2x_2 + 2\sqrt{x_1 x_2}}
$


$x_1 \frac{\partial f}{\partial x_1} +  x_2 \frac{\partial f}{\partial x_2} = \frac{1}{2}$

$x_1 \frac{1}{ 2x_1 + 2\sqrt{x_1 x_2}} + x_2 \frac{1}{ 2x_2 + 2\sqrt{x_1 x_2}} = \frac{1}{2}$

$\frac{1}{2}x_1 \frac{1}{ x_1 + \sqrt{x_1 x_2}} + \frac{1}{2}x_2 \frac{1}{ x_2 + \sqrt{x_1 x_2}} = \frac{1}{2}$

$x_1 \frac{1}{x_1 + \sqrt{x_1 x_2}} + x_2 \frac{1}{x_2 + \sqrt{x_1 x_2}} = 1$

$x_1 \frac{x_1 - \sqrt{x_1 x_2}}{(x_1 - \sqrt{x_1 x_2})(x_1 + \sqrt{x_1 x_2})} + x_2 \frac{x_2 - \sqrt{x_1 x_2}}{(x_2 - \sqrt{x_1 x_2})(x_2 + \sqrt{x_1 x_2})} = 1$

$x_1 \frac{x_1 - \sqrt{x_1 x_2}}{x_1^2 - (x_1 x_2)^2} + x_2 \frac{x_2 - \sqrt{x_1 x_2}}{x_2^2 - (x_1 x_2)^2} = 1$

Предположим $x_1 = 0$. Тогда:

$x_1 \frac{x_1 - \sqrt{x_1 x_2}}{x_1^2 - (x_1 x_2)^2} + x_2 \frac{x_2 - \sqrt{x_1 x_2}}{x_2^2 - (x_1 x_2)^2} = 1$

$0 + x_2 \frac{x_2 - 0}{x_2^2 - 0} = 1$

$\frac{x_2 \cdot x_2}{x_2^2} = 1$

$\frac{x_2^2}{x_2^2} = 1$

$\frac{1}{1} = 1$

При $x_2 = 0$ ситуация аналогична:

$x_1 \frac{x_1 - \sqrt{x_1 x_2}}{x_1^2 - (x_1 x_2)^2} + x_2 \frac{x_2 - \sqrt{x_1 x_2}}{x_2^2 - (x_1 x_2)^2} = 1$

$x_1 \frac{x_1 - 0}{x_1^2 - 0} + 0 = 1$

$\frac{x_1 \cdot x_1}{x_1^2} = 1$

$\frac{x_1^2}{x_1^2} = 1$

$\frac{1}{1} = 1$

Можно сделать вывод, что функция $f(x) = \ln(\sqrt{x_1} + \sqrt{x_2})$ удовлетворяет условию $x_1 \frac{\partial f}{\partial x_1} +  x_2 \frac{\partial f}{\partial x_2} = \frac{1}{2}$ при $\begin{cases} x_1 = 0, \forall x_2 \in \mathbb{R} \\ x_2 = 0, \forall x_1 \in \mathbb{R} \end{cases}$

#### Ответ:

$\begin{cases} x_1 = 0, \forall x_2 \in \mathbb{R} \\ x_2 = 0, \forall x_1 \in \mathbb{R} \end{cases}$
