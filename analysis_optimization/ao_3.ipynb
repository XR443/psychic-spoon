#### Дано:

$f : \mathbb{R}^3 → \mathbb{R}^2$

$f(x, y, z) = (x + y + z, xyz)^\top$

#### Найти:

Матрицу Якоби $J_f$ функции $f$

Значение в точке $v = \begin{pmatrix} 1 & 2 & 3 \end{pmatrix}^\top$

#### Решение

$
J(x, y, z) = 
\begin{pmatrix}
\frac{\partial F_1}{\partial x} & \frac{\partial F_1}{\partial y} & \frac{\partial F_1}{\partial z} \\ 
\frac{\partial F_2}{\partial x} & \frac{\partial F_2}{\partial y} & \frac{\partial F_2}{\partial z} \\ 
\end{pmatrix}
= \begin{pmatrix}
1 & 1 & 1 \\ 
yz & xz & xy \\ 
\end{pmatrix}
$

$\frac{\partial F_1}{\partial x} = (x + y + z)' = 1$

$\frac{\partial F_2}{\partial x} = (xyz)' = yz$

$\frac{\partial F_1}{\partial y} = (x + y + z)' = 1$

$\frac{\partial F_2}{\partial y} = (xyz)' = xz$

$\frac{\partial F_1}{\partial z} = (x + y + z)' = 1$

$\frac{\partial F_2}{\partial z} = (xyz)' = xy$

Для $v = \begin{pmatrix} 1 & 2 & 3 \end{pmatrix}^\top$ $J_f$ принимает вид:

$
J(v) 
= \begin{pmatrix}
1 & 1 & 1 \\ 
yz & xz & xy \\ 
\end{pmatrix}
= \begin{pmatrix}
1 & 1 & 1 \\ 
2 \cdot 3 & 1 \cdot 3 & 1 \cdot 2 \\ 
\end{pmatrix}
= \begin{pmatrix}
1 & 1 & 1 \\ 
6 & 3 & 2 \\ 
\end{pmatrix}
$

#### Ответ:

$
J(v) = \begin{pmatrix}
1 & 1 & 1 \\ 
yz & xz & xy \\ 
\end{pmatrix}
$

Для $v = \begin{pmatrix} 1 & 2 & 3 \end{pmatrix}^\top$,
$
J(v) = \begin{pmatrix}
1 & 1 & 1 \\ 
6 & 3 & 2 \\ 
\end{pmatrix}
$
