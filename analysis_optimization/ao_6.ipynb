#### Дано:

$f(x) = \begin{Vmatrix} Ax \end{Vmatrix}_2$, $x \in \mathbb{R}^n \backslash \{0\}, A \in \mathbb{R}^{m \times n}$

#### Найти:

$df(x)$

$\nabla f(x)$

#### Решение

$f(x) = \begin{Vmatrix} Ax \end{Vmatrix}_2$

$\begin{Vmatrix} Ax \end{Vmatrix}_2 = \sqrt{\sum_i |Ax|^2} = \sqrt{<Ax, Ax>}$

$\sqrt{<Ax, Ax>} \rightarrow min$

$
dF(x) = \sqrt{<Ax, Ax>} 
= d(<Ax, Ax>^\frac{1}{2}) \\
= \frac{1}{2} \frac{1}{<Ax, Ax>^\frac{1}{2}} d(<Ax,Ax>) \\
= \frac{1}{2} \frac{1}{\sqrt{<Ax, Ax>}} d(<Ax,Ax>) \\
= \frac{1}{2} \frac{1}{\sqrt{<Ax, Ax>}} \cdot (<d(Ax),Ax> + <Ax,d(Ax)>) \\
= \frac{1}{2} \frac{1}{\sqrt{<Ax, Ax>}} \cdot (<d(x),A^\top \cdot Ax> + <A^\top \cdot Ax,d(x)>) \\
= \frac{1}{2} \frac{1}{\sqrt{<Ax, Ax>}} \cdot (<d(x),A^\top \cdot Ax> + <A^\top \cdot Ax,d(x)>) \\
= \frac{1}{2} \frac{1}{\sqrt{<Ax, Ax>}} \cdot (<d(x),Ex> + <Ex,d(x)>) \\
= \frac{1}{2} \frac{1}{\sqrt{<Ax, Ax>}} \cdot 2<x, dx> \\
= \frac{1}{\sqrt{<Ax, Ax>}} \cdot <x, dx> \\
= \frac{<x, dx>}{\sqrt{<Ax, Ax>}} \\
= \frac{x}{\sqrt{<Ax, Ax>}} \\
= \frac{x}{{||Ax||}_2} \\
$

#### Ответ:

$\frac{x}{{||Ax||}_2}$
