### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
rg = np.random
arr = rg.random((4, 3, 2))
(arr.min(), arr.max())
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
A = rg.random((6, 4))
B = rg.random((4, 3))
A @ B # A.dot(B)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
A = rg.randint(1, 100, (7, 7))
norm = 1 / A.max()
(A.mean(), A.std(), A * norm)
```

```python
tips[tips["smoker"]== "Yes"]["smoker"].count()
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
tips.groupby("day")["total_bill"].mean()
```

```python
tips.groupby("day")["tip"].mean()
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
median_value = tips["total_bill"].median()

tips[tips["total_bill"] > median_value].groupby("sex")["tip"].mean()
```

```python
tips.groupby("sex")["tip"].mean()
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
tips["smoker"] = tips["smoker"].apply(lambda x: x == "Yes")
tips
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
tips["total_bill"].plot.hist()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
tips.plot.scatter(x="total_bill", y="tip")
```

### Постройте pairplot

```python
seaborn.pairplot(tips)
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
tips[["total_bill", "day"]].groupby("day").mean().plot()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
tips[["time", "tip"]].plot.hist(column="tip", by="time")
```

```python
tips[["time", "total_bill"]].plot.hist(column="total_bill", by="time")
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
tips[tips["sex"] == "Male"][["total_bill", "tip"]].plot.scatter(x="total_bill", y="tip", c="#4682B4")
```

```python
tips[tips["sex"] == "Female"][["total_bill", "tip"]].plot.scatter(x="total_bill", y="tip", c="#ff7052")
```

```python
tips.groupby("sex").count()
```

```python
bill_and_tips = tips[["total_bill", "tip"]]
bill_and_tips["total_bill_group"] = bill_and_tips["total_bill"].apply(lambda x: x // 5)
bill_and_tips.groupby("total_bill_group").mean()[["total_bill", "tip"]].plot(x="total_bill", y="tip")
```

```python
tips_percent = tips[["total_bill", "tip"]]
tips_percent["tips_percent"] = tips["tip"] / tips["total_bill"] * 100
tips_percent
```
